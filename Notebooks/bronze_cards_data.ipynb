```python
## Importando bibliotecas
import pandas as pd
from pyspark.sql.functions import current_date, current_timestamp, expr
```

```python
## Criando Variáveis
database = "bronze"
tabela = "cards_data"
path_arquivo = "pnbank.bronze.cards_data"
df = spark.table(path_arquivo)
```

```python
## Adicionando colunas de controle
df = df.withColumn("data_carga", current_date())
df = df.withColumn("data_hora_carga", expr("current_timestamp() - INTERVAL 3 HOURS"))
```

```python
## Salvando dados no Data Lake
df.write \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .option("overwriteSchema", "true")\
    .saveAsTable("pnbank.bronze.cards_data")

print("Dados gravados com sucesso!")
```
